```python
def _to_numpy(x):
    if x is None:
        return None
    if isinstance(x, np.ndarray):
        return x
    return x.detach().cpu().numpy()

def _periodic_positions(positions, domain):
    min_d = domain.min
    max_d = domain.max
    periodicity = domain.periodic
    wrapped = [
        (torch.remainder(positions[:, i] - min_d[i], max_d[i] - min_d[i]) + min_d[i]) if periodicity[i] else positions[:, i]
        for i in range(domain.dim)
    ]
    return torch.stack(wrapped, dim=-1)

def _pad_positions_for_fastplotlib(value):
    if value is None:
        return None
    if value.ndim == 2 and value.shape[1] == 2:
        return np.pad(value, ((0, 0), (0, 1)), mode='constant')
    return value

def _scalar_to_rgba(values, cmap_name='viridis'):
    values = np.asarray(values, dtype=np.float32)
    if values.size == 0:
        return np.zeros((0, 4), dtype=np.float32)
    vmin = float(np.nanmin(values))
    vmax = float(np.nanmax(values))
    if not np.isfinite(vmin) or not np.isfinite(vmax):
        vmin, vmax = 0.0, 1.0
    if np.isclose(vmin, vmax):
        vmax = vmin + 1e-6
    norm = plt.Normalize(vmin=vmin, vmax=vmax)
    rgba = plt.get_cmap(cmap_name)(norm(values)).astype(np.float32)
    return rgba

def _update_feature_value(feature, value):
    if feature is None or value is None:
        return False
    try:
        feature[:] = value
        return True
    except Exception:
        pass
    try:
        feature.__setitem__(slice(None), value)
        return True
    except Exception:
        pass
    if hasattr(feature, 'set_value'):
        try:
            feature.set_value(value)
            return True
        except TypeError:
            try:
                feature.set_value(feature, value)
                return True
            except Exception:
                pass
        except Exception:
            pass
    if hasattr(feature, 'value'):
        try:
            feature.value = value
            return True
        except Exception:
            pass
    return False

def _safe_update_graphic_data(graphic, value):
    if value is None:
        return
    patched_value = value
    if hasattr(graphic, 'data'):
        data_feature = graphic.data
        if data_feature.__class__.__name__ == 'VertexPositions':
            patched_value = _pad_positions_for_fastplotlib(value)
        if _update_feature_value(data_feature, patched_value):
            return
    if hasattr(graphic, '_data') and _update_feature_value(graphic._data, patched_value):
        return
    if hasattr(graphic, 'set_data'):
        try:
            graphic.set_data(patched_value)
            return
        except Exception:
            pass
    raise RuntimeError('Could not update graphic data for fastplotlib object')

def _safe_update_graphic_scalar_colors(graphic, values, cmap_name='viridis'):
    if values is None:
        return
    rgba = _scalar_to_rgba(values, cmap_name=cmap_name)
    if hasattr(graphic, 'colors') and _update_feature_value(graphic.colors, rgba):
        return
    if hasattr(graphic, '_colors') and _update_feature_value(graphic._colors, rgba):
        return
    raise RuntimeError('Could not update graphic colors for fastplotlib object')

def _refresh_fastplotlib_ranges(viz):
    if viz['backend'] != 'fastplotlib':
        return
    if 'pressure' in viz and hasattr(viz['pressure'], 'reset_vmin_vmax'):
        viz['pressure'].reset_vmin_vmax()

def _request_fastplotlib_redraw(viz):
    if viz['backend'] != 'fastplotlib':
        return
    fig = viz.get('fig')
    if fig is not None and hasattr(fig, 'canvas') and fig.canvas is not None:
        try:
            fig.canvas.request_draw()
            return
        except Exception:
            pass
        try:
            fig.canvas._rc_request_draw()
            return
        except Exception:
            pass

def _fluid_mask(system_state):
    if hasattr(system_state, 'kinds'):
        return system_state.kinds == 0
    return torch.ones(system_state.positions.shape[0], dtype=torch.bool, device=system_state.positions.device)

def _fluid_particles(system_state):
    filtered = filterParticles(system_state, which='fluid')
    if filtered is None:
        raise ValueError('No fluid particles available for visualization')
    return filtered

def setup_runtime_visualization(particleSystem, config, initialEnergy, simulationName, dt, marker_size=1):
    if PLOT_BACKEND == 'fastplotlib' and FASTPLOTLIB_AVAILABLE:
        fig = fpl.Figure(shape=(1, 2), size=(1200, 500), canvas='jupyter')
        a0 = fig[0, 0]
        a1 = fig[0, 1]
        
        # Set axis titles
        a0.title = "Velocity Magnitude"
        a1.title = "Pressure (mean-subtracted)"

        fluid_particles = _fluid_particles(particleSystem.systemState)
        fluid_mask = _fluid_mask(particleSystem.systemState)
        fluid_positions = _periodic_positions(fluid_particles.positions, config['domain'])
        fluid_vel = torch.linalg.norm(particleSystem.systemState.velocities[fluid_mask], dim=-1)
        velocity_values = _to_numpy(fluid_vel).astype(np.float32)

        velocity_scatter = a0.add_scatter(
            data=_pad_positions_for_fastplotlib(_to_numpy(fluid_positions[:, :2]).astype(np.float32)),
            colors=_scalar_to_rgba(velocity_values, cmap_name='viridis'),
            sizes=float(marker_size),
            mode='simple'
        )
        
        # Add colorbar info text for velocity field
        vel_min, vel_max = np.nanmin(velocity_values), np.nanmax(velocity_values)
        # a0.add_text(
        #     text=f"Min: {vel_min:.3f}\nMax: {vel_max:.3f}",
        #     anchor=(0.02, 0.98),
        #     size=10,
        #     color="white"
        # )

        pressure = particleSystem.systemState.pressures[fluid_mask] - particleSystem.systemState.pressures[fluid_mask].mean()
        grid, pressure_grid = mapToGrid(
            fluid_particles,
            pressure,
            config['domain'],
            config['kernel'],
            128,
        )
        
        # Create image and position it with physical coordinates
        pressure_img = a1.add_image(data=_to_numpy(pressure_grid.T).astype(np.float32), cmap='RdBu_r')
        if hasattr(pressure_img, 'reset_vmin_vmax'):
            pressure_img.reset_vmin_vmax()
        
        # Position image at domain center and set physical extent
        # grid is a tuple of (grid_x, grid_y, ...) from torch.meshgrid
        domain_min = _to_numpy(config['domain'].min)[:2]  # Get first 2D coordinates
        domain_max = _to_numpy(config['domain'].max)[:2]
        domain_center = (domain_min + domain_max) / 2
        
        # Set image offset to domain center for proper physical coordinate alignment
        pressure_img.offset = np.array([domain_center[0], domain_center[1], 0], dtype=np.float32)
        
        # Add colorbar info text for pressure field
        pressure_values = _to_numpy(pressure_grid).astype(np.float32)
        p_min, p_max = np.nanmin(pressure_values), np.nanmax(pressure_values)
        # a1.add_text(
        #     text=f"Min: {p_min:.3f}\nMax: {p_max:.3f}",
        #     anchor=(0.02, 0.98),
        #     size=10,
        #     color="white"
        # )

        if hasattr(fig, 'canvas') and fig.canvas is not None:
            fig.canvas.css_width = '1200px'
            fig.canvas.css_height = '500px'
            fig.canvas.set_logical_size(1200, 500)

        widget = fig.show()
        display(widget)
        return fig, {
            'backend': 'fastplotlib',
            'fig': fig,
            'widget': widget,
            'velocity': velocity_scatter,
            'pressure': pressure_img,
            'velocity_cmap': 'viridis',
            'domain_center': domain_center,
            'domain_min': domain_min,
            'domain_max': domain_max,
        }

    fig, axis = plt.subplots(1, 2, figsize=(12, 5), squeeze=False, sharey=True)

    velocityPlot = visualizeParticles(
        fig, axis[0, 0],
        particles=particleSystem.systemState,
        domain=config['domain'],
        quantity=particleSystem.systemState.velocities,
        which='fluid',
        mapping='L2',
        cmap='viridis',
        visualizeBoth=False,
        kernel=config['kernel'],
        plotDomain=False,
        gridVisualization=False, markerSize=1, streamLines=False
    )

    pressurePlot = visualizeParticles(
        fig, axis[0, 1],
        particles=particleSystem.systemState,
        domain=config['domain'],
        quantity=particleSystem.systemState.pressures - particleSystem.systemState.pressures.mean(),
        which='fluid',
        mapping='.x',
        cmap='RdBu_r',
        visualizeBoth=False,
        kernel=config['kernel'],
        plotDomain=False,
        gridVisualization=True, markerSize=3, streamLines=False
    )

    axis[0, 0].set_title('Velocity')
    axis[0, 1].set_title('Pressure (mean subtracted)')

    rhoMin = particleSystem.systemState.densities.min().detach().cpu().item() / config['fluid']['rho0']
    rhoMean = particleSystem.systemState.densities.mean().detach().cpu().item() / config['fluid']['rho0']
    rhoMax = particleSystem.systemState.densities.max().detach().cpu().item() / config['fluid']['rho0']

    kineticEnergy = 0.5 * particleSystem.systemState.masses / config['fluid']['rho0'] * particleSystem.systemState.densities * torch.linalg.norm(particleSystem.systemState.velocities, dim=-1) ** 2
    potentialEnergy = None
    if config['gravity']['active'] and 'mode' in config['gravity'] and config['gravity']['mode'] == 'potential':
        B = config['gravity']['magnitude']
        potentialEnergy = 0.5 * B ** 2 * particleSystem.systemState.masses / config['fluid']['rho0'] * particleSystem.systemState.densities * torch.linalg.norm(particleSystem.systemState.positions, dim=-1) ** 2

    combinedEnergy = kineticEnergy + potentialEnergy if potentialEnergy is not None else kineticEnergy
    totalInitialEnergy = (initialEnergy).sum().detach().cpu().item()
    totalEnergy = (combinedEnergy).sum().detach().cpu().item()

    fig.suptitle(
        f'{simulationName}, ptcls = {particleSystem.systemState.positions.shape[0]}, kernel = {config["kernel"].name}, neighbors = {config["targetNeighbors"]:.2g}, $c_s$ = {config["fluid"]["c_s"]:.1f}\n$\\rho$ = [{rhoMin:.4g} | {rhoMean:.4g} | {rhoMax:.4g}], $E_0$ = {totalInitialEnergy:.4g}, $E$ = {totalEnergy:.4g}, $\\Delta E$ = {totalEnergy - totalInitialEnergy:.4g}, $t$ = {particleSystem.t:.4g}, $\\Delta t$ = {dt:.2e}'
    )
    fig.tight_layout()

    return fig, {
        'backend': 'matplotlib',
        'fig': fig,
        'axis': axis,
        'velocity': velocityPlot,
        'pressure': pressurePlot,
    }

def update_runtime_visualization(viz, particleSystem, config):
    if viz['backend'] == 'fastplotlib':
        fluid_particles = _fluid_particles(particleSystem.systemState)
        fluid_mask = _fluid_mask(particleSystem.systemState)
        fluid_positions = _periodic_positions(fluid_particles.positions, config['domain'])
        fluid_vel = torch.linalg.norm(particleSystem.systemState.velocities[fluid_mask], dim=-1)

        _safe_update_graphic_data(viz['velocity'], _to_numpy(fluid_positions[:, :2]).astype(np.float32))
        velocity_values = _to_numpy(fluid_vel).astype(np.float32)
        _safe_update_graphic_scalar_colors(viz['velocity'], velocity_values, cmap_name=viz.get('velocity_cmap', 'viridis'))
        
        # Update velocity colorbar text
        vel_min, vel_max = np.nanmin(velocity_values), np.nanmax(velocity_values)
        # vel_axis = viz['velocity'].parent
        # if hasattr(vel_axis, 'graphics'):
        #     text_graphics = [g for g in vel_axis.graphics if hasattr(g, '__class__') and 'text' in g.__class__.__name__.lower()]
        #     if text_graphics:
        #         # Update the text content (this is a simplified approach)
        #         pass

        pressure = particleSystem.systemState.pressures[fluid_mask] - particleSystem.systemState.pressures[fluid_mask].mean()
        _, pressure_grid = mapToGrid(
            fluid_particles,
            pressure,
            config['domain'],
            config['kernel'],
            128,
        )
        _safe_update_graphic_data(viz['pressure'], _to_numpy(pressure_grid.T).astype(np.float32))
        
        # Restore pressure image offset to maintain physical coordinate alignment
        if 'domain_center' in viz:
            viz['pressure'].offset = np.array([viz['domain_center'][0], viz['domain_center'][1], 0], dtype=np.float32)
        
        _refresh_fastplotlib_ranges(viz)
        _request_fastplotlib_redraw(viz)
        return

    updatePlot(viz['velocity'], particleSystem.systemState, particleSystem.systemState.velocities)
    updatePlot(viz['pressure'], particleSystem.systemState, particleSystem.systemState.pressures - particleSystem.systemState.pressures.mean())
```
